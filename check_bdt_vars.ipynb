```python
def plot_var_presel_breakdown(var,nbins,min_bin,max_bin,force_legend='',logy=False,add_query=''):

    bins = np.linspace(min_bin, max_bin, num=nbins+1)

    query = "ssm_kine_energy>0"
    if add_query!='': query += ' and ' + add_query 

    lowE_query
    anti_lowE_query = "not ("+lowE_query+")"
      
    y_kdar_test,x = np.histogram(kdar_overlay_df.query(query).query(kdar.presel_query)[var].to_numpy(),bins=bins,density=False,weights=kdar_overlay_df.query(query).query(kdar.presel_query)["net_weight"].to_numpy()*kdar_scale_test*kdar_xs_scale)
    y_lowE_test,x = np.histogram(overlay_df.query(lowE_query).query(query).query(kdar.presel_query)[var].to_numpy(),bins=bins,density=False,weights=overlay_df.query(lowE_query).query(query).query(kdar.presel_query)["net_weight"].to_numpy()*overlay_scale_test*lowE_xs_scale)
    y_overlay_test,x = np.histogram(overlay_df.query(anti_lowE_query).query(query).query(kdar.presel_query)[var].to_numpy(),bins=bins,density=False,weights=overlay_df.query(anti_lowE_query).query(query).query(kdar.presel_query)["net_weight"].to_numpy()*overlay_scale_test*overlay_xs_scale)
    #y_dirt_test,x = np.histogram(dirt_df.query(query).query(kdar.presel_query)[var].to_numpy(),bins=bins,density=False,weights=dirt_df.query(query).query(kdar.presel_query)["net_weight"].to_numpy()*dirt_scale_test*dirt_xs_scale)
   
    y_ext_test,x = np.histogram(ext_df.query(query).query(kdar.presel_query)[var].to_numpy(),bins=bins,density=False,weights=np.ones_like(ext_df.query(query).query(kdar.presel_query)[var].to_numpy())*ext_scale_test*ext_xs_scale)
    y_test = y_kdar_test+y_lowE_test+y_overlay_test+y_dirt_test+y_ext_test
    
    y_data,x = np.histogram(data_df.query(query).query(kdar.presel_query)[var].to_numpy(),bins=bins,density=False)
    
    plt.figure()      
    #plt.stairs(y_kdar_test+y_lowE_test+y_overlay_test+y_ext_test+y_dirt_test,x,baseline=y_kdar_test+y_lowE_test+y_overlay_test+y_ext_test,fill=True,color="violet",label="DIRT",alpha=0.5)
    plt.stairs(y_kdar_test+y_lowE_test+y_overlay_test+y_ext_test,x,baseline=y_kdar_test+y_lowE_test+y_overlay_test,fill=True,color="brown",label="EXT",alpha=0.5)
    plt.stairs(y_kdar_test+y_lowE_test+y_overlay_test,x,baseline=y_kdar_test+y_lowE_test,fill=True,color="green",label=f"other $\nu$",alpha=0.5) 
    plt.stairs(y_kdar_test+y_lowE_test,x,baseline=y_kdar_test,fill=True,color="red",label=r"$\nu_\mu/\bar{\nu}_\mu$<400 MeV",alpha=0.5)       
    plt.stairs(y_kdar_test,x,fill=True,color="dodgerblue",label="KDAR",alpha=0.5)    
    plt.errorbar(utils.get_bin_centers(x),y_data,yerr=np.sqrt(y_data),ms=8, lw=1,fmt='.',ecolor = 'black',color='black', capsize=2, capthick=1, label="Data")
    if force_legend=='outside': plt.legend(loc='center left', bbox_to_anchor=(1, 0.5)) 
    elif force_legend !='' and force_legend !='off': plt.legend(loc=force_legend)
    elif force_legend !='off': plt.legend()
    if logy: plt.yscale()
    plt.xlim(min_bin,max_bin)
    plt.xlabel(var)
    plt.ylabel("Counts")
    plt.show()
```
